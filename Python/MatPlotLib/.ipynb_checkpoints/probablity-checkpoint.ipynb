```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
data = np.random.uniform(0,1, 10000)
```

```python
plt.hist(data, color="lightgreen", density=True, alpha=0.6, bins=90)
plt.xlabel("Value")
plt.ylabel("Probablity Density")
plt.title("Probablity Mass Function")
plt.grid=True
```
