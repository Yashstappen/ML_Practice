```python
from sklearn.tree import DecisionTreeClassifier
```

```python
# Features: [weight (grams), color (0 = red, 1 = orange)]
features = [
    [150, 0],  # Apple
    [170, 0],  # Apple
    [130, 1],  # Orange
    [120, 1],  # Orange
]
```

```python
labels = ["apple", "apple", "orange", "orange"]
```

```python
#Training the model
clf = DecisionTreeClassifier()
clf = clf.fit(features, labels)
```

```python
#Prediction
prediction = clf.predict([[145, 1]])
print(prediction)
```

```python
from sklearn.tree import DecisionTreeClassifier
import random
from sklearn import tree
import matplotlib.pyplot as plt

#features
# [No_of_hours_studied, No_of_hours_slept]
feature=[]
label=[]

#DataSet
for i in range(50):
    
    study = random.randint(1,10)
    sleep = random.randint(4,8)

    if(study>=5 and sleep>=6):
        result = "pass"
    else:
        result = "fail"

    feature.append([study, sleep])
    label.append(result)
#Seeing the dataset
print(feature[:5])
print(label[:5])

#TrainingTheModel
model = DecisionTreeClassifier()
model.fit(feature, label)

#Predicting the output
x = int(input("Enter the number of hours studied: "))
y = int(input("Enter the number of hours slept: "))
prediction = model.predict([[x, y]])
print(prediction)

#Visualizing the working
tree.plot_tree(model, filled=True)
plt.show()
```
